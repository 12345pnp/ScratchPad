```python
def levenshtein_distance(source: str, target: str) -> int:
    source_len, target_len = len(source), len(target)
    
    levenshtein_matrix = [[0]*(target_len + 1) for _ in range(source_len + 1)]
    
    for j in range(1, target_len + 1):
        levenshtein_matrix[0][j] = j
    
    for i in range(1, source_len + 1):
        levenshtein_matrix[i][0] = i
    
    for i in range(1, source_len + 1):
        for j in range(1, target_len + 1):
            levenshtein_matrix[i][j] = min(
                levenshtein_matrix[i - 1][j] + 1,
                levenshtein_matrix[i][j - 1] + 1,
                levenshtein_matrix[i - 1][j - 1] + (1 if source[i - 1] != target[j - 1] else 0)
            )
    return levenshtein_matrix[-1][-1]
```

```python
%%cython
import numpy as np

def levenshtein_distance_cython(str source, str target) -> int:
    cdef int source_len = len(source)
    cdef int target_len = len(target)
    
    cdef long[:, ::1] levenshtein_matrix = np.zeros((source_len + 1, target_len + 1), dtype=int)

    cdef int i = 0
    cdef int j = 0
    
    for j in range(1, target_len + 1):
        levenshtein_matrix[0, j] = j
    
    for i in range(1, source_len + 1):
        levenshtein_matrix[i, 0] = i
    
    for i in range(1, source_len + 1):
        for j in range(1, target_len + 1):
            levenshtein_matrix[i, j] = min(
                levenshtein_matrix[i - 1, j] + 1,
                levenshtein_matrix[i, j - 1] + 1,
                levenshtein_matrix[i - 1, j - 1] + (1 if source[i - 1] != target[j - 1] else 0)
            )
    return levenshtein_matrix[-1, -1]


```
